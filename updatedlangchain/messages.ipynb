# Messages

```python
from langchain_google_genai import ChatGoogleGenerativeAI

google_gen_ai_model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

## Text Prompts

```python
google_gen_ai_model.invoke("What is Langchain?").text
```

## Message Prompt

- **System Message** – Tells the model how to behave and provides context for interactions.
- **Human Message** – Represents user input and interactions with the model.
- **AI Message** – Responses generated by the model, including text content, tool calls, and metadata.
- **Tool Message** – Represents the outputs of tool calls.

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are a historian"),
    HumanMessage("Tell concisely how Stalin spread the communism over the world")
]
```

```python
google_gen_ai_model.invoke(messages).text
```

```python
system_msg = SystemMessage("You are a helpful coding assistant.")

messages = [
    system_msg,
    HumanMessage("How do i create a Rest API")
]
response = google_gen_ai_model.invoke(messages)
print(response.text)
```

```python
for chunk in google_gen_ai_model.stream(messages):
    print(chunk.text, end="|",flush=True)
```

```python
human_msg = HumanMessage(
    content="Hello!",
    name="Ravi",
    id="528"
)

google_gen_ai_model.invoke([human_msg]).text
```

```python
from langchain.messages import ToolMessage, AIMessage

ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name":"get_weather",
        "args":{"location":"San Francisco"},
        "id":"call_123"
    }]
)

weather_result = "Sunny, 72F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"
)

messages = [
    HumanMessage("What's the weather in a location?"),
    ai_message,
    tool_message
]

response = google_gen_ai_model.invoke(messages)
```
